```python
import pandas as pd
import sqlalchemy as sa
from sqlalchemy import create_engine, MetaData, Table, inspect, text
engine = create_engine('sqlite:///:memory:')
```

```python
banco = pd.read_csv(r'C:\Users\JulianoGarciaGregóri\OneDrive - PerformaIT Cons. e Des. de Softwares\Documentos\estudos-python\Pandas-IO\Aula05\clientes_banco.csv')
banco.head()
```

```python
banco.to_sql('clientes', con=engine, index=False)
```

```python
inspector = inspect(engine)
inspector.get_table_names()
```

```python
query = 'SELECT * FROM clientes WHERE Categoria_de_renda == "Empregado"'
resultado = pd.read_sql_query(query, con=engine)
resultado.head()
```

```python
resultado.to_sql('clientes_empregados', con=engine, index=False)    
```
